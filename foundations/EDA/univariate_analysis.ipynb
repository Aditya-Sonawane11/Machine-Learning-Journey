```python
import pandas as pd
import matplotlib.pyplot as plt

housing = pd.read_csv("housing.csv")
```

```python
numerical_cols = [
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "median_house_value"
]

categorical_cols = [
    "ocean_proximity"
]
```

```python
housing[numerical_cols].describe()
```

```python
housing[numerical_cols].median()
```

```python
housing[numerical_cols].mode().iloc[0]
```

```python
housing[numerical_cols].var()
```

```python
housing[numerical_cols].skew()
```

```python
housing[numerical_cols].skew().sort_values()
```

```python
housing[numerical_cols].kurtosis()
```

```python
housing[numerical_cols].hist(
    bins=40,
    figsize=(16, 12),
    edgecolor="black"
)

plt.suptitle(
    "Univariate Analysis: Numerical Distributions",
    fontsize=16
)
plt.tight_layout()
plt.show()
```

```python
for col in numerical_cols:
    values = housing[col].dropna()

    plt.figure(figsize=(10, 2.5))
    plt.boxplot(values, vert=False)

    plt.title(f"Box Plot: {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

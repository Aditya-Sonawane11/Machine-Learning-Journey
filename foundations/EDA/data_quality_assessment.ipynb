    Target: median_house_value, a numerical variable.<br><br>

    The initial check found zero exact duplicate rows. This is useful, although it doesn't guarantee there are no other data-quality problems.<br><br>
    <b>6. Model feasibility</b><br>
    Linear regression is a reasonable first baseline. The categorical feature will need suitable encoding, and the numerical features may benefit from further investigation and preprocessing.<br><br>
    <b>7. Remaining uncertainty</b><br>
    We haven't yet established the distributions of individual features, the presence of influential observations, relationships with house value, or whether the data contains unusual target values. Those questions belong to subsequent EDA stages.

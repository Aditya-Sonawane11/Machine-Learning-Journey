<h3 align="center" style="color:rgb(1, 126, 250);"><b>STANDARDIZATION</b></h3>

`Standardization is a feature-scaling technique that transforms numerical data so that it has a mean of approximately 0 and a standard deviation of 1.`


> **In a nutshell:** Standardization rescales numerical features so they have a mean of approximately **0** and a standard deviation of **1**.

---

#### 1. Formula

$$
z = \frac{x - \mu}{\sigma}
$$

| Symbol | Meaning |
|---|---|
| \(x\) | Original value |
| \(\mu\) | Feature mean |
| \(\sigma\) | Feature standard deviation |
| \(z\) | Standardized value |

**Example**

If $$(x = 8.3), (mu = 5), and (sigma = 2):$$

$$
z = \frac{8.3 - 5}{2} = 1.65
$$

The value is **1.65 standard deviations above the mean**.

---

#### 2. Why Use It?

- Makes features with different scales comparable.
- Often useful for **KNN, SVM, Logistic Regression, and PCA**.
- Usually unnecessary for **Decision Trees and Random Forests**.

---

#### 3. Python with Scikit-learn

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

- `fit()` learns the mean and standard deviation.
- `transform()` scales data using those learned values.
- `fit_transform()` does both in one step.

---

#### 4. Standardization vs. Min-Max Scaling

| Standardization | Min-Max Scaling |
|---|---|
| Mean becomes approximately `0` | Usually maps values to `0–1` |
| Standard deviation becomes approximately `1` | Does not set a target standard deviation |
| Uses mean and standard deviation | Uses minimum and maximum |

---

#### 5. Important Reminders

- Standardized values can be negative, zero, or positive.
- Values are **not restricted to `0–1`**.
- Standardization does **not** make data normally distributed.
- It is sensitive to outliers.
- **Avoid data leakage:** Fit the scaler on training data only, then use it to transform both training and test data.

---

##### Quick Revision

**Formula:**$$(z = (x - mu) / sigma)$$  
**Mean after scaling:** approximately `0`  
**Standard deviation:** approximately `1`  
**Scikit-learn:** `StandardScaler`  
**Golden rule:** Fit on training data; transform test data.


````python
import pandas as pd
from sklearn.preprocessing import StandardScaler
````

````python
df = pd.read_csv("../00_datasets/housing.csv")

print(df.head())
print(df[["median_income", "total_rooms", "population"]].describe())
````

````python
scaler = StandardScaler()

columns = ["median_income", "total_rooms", "population"]

scaled_data = scaler.fit_transform(df[columns])

scaled_df = pd.DataFrame(
    scaled_data,
    columns=columns,
    index=df.index
)

print(scaled_df.head())
````

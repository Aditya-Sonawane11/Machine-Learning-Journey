<h3 align="center" style="color:rgb(1, 126, 250);"><b>NORMALIZATION (Min - Max Scaling)</b></h3>

> **Goal:** Rescale numerical features to a fixed range, usually `0` to `1`.

##### Formula

$$X' = \frac{X-X_{\min}}{X_{\max}-X_{\min}}$$

##### Scikit-learn

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)
```

##### When to Use
- Features have different ranges.
- A fixed range such as `0–1` is desirable.
- Common examples: **KNN, Neural Networks, distance-based algorithms**.

##### Remember
- Sensitive to outliers because extreme values affect the minimum and maximum.
- Does not make data normally distributed.
- Fit on training data only; transform test data using the same scaler.
- Unlike standardization, it does not aim for mean `0` and standard deviation `1`.


````python
import pandas as pd 
from sklearn.preprocessing import MinMaxScaler
````

````python
df = pd.read_csv("../00_datasets/housing.csv")
````

````python
df.sample(5)
````

````python
columns = ["median_income", "total_rooms", "population"]

X = df[columns]

X.head()
````

````python
scaler = MinMaxScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=columns,
    index=X.index
)

X_scaled.head()
````

````python
scaler.fit_transform(X)
````

````python
print("Original data:")
print(X.head())
````

````python
print("\nNormalized data:")
print(X_scaled.head())
````

````python
print("Minimum values learned:", scaler.data_min_)
print("Maximum values learned:", scaler.data_max_)
print("Feature ranges:", scaler.data_range_)
````

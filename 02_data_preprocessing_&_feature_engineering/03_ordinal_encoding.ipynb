<h3 align="center" style="color:rgb(1, 126, 250);"><b>ORDINAL  ENCODING</b></h3>

> **Goal:** Convert categorical values into numerical values while preserving a meaningful order between categories.

##### Example

| Category | Encoded Value |
|---|---:|
| Low | 0 |
| Medium | 1 |
| High | 2 |

##### Scikit-learn

```python
from sklearn.preprocessing import OrdinalEncoder

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

X_encoded = encoder.fit_transform(X)
```

##### When to Use
- Categories have a meaningful ranking.
- Examples: Education level, satisfaction ratings, size categories.

##### Remember
- Define the category order explicitly when it matters.
- Do not assign an artificial ranking to nominal categories.
- `OrdinalEncoder` is for input features (`X`).
- `LabelEncoder` is primarily for target labels (`y`), such as class names.
- Integer encoding does not automatically make a model understand the intended ranking.


````python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.read_csv("../00_datasets/housing.csv")

df.head()
````

````python
df["income_level"] = pd.cut(
    df["median_income"],
    bins=[-float("inf"), 2, 4, float("inf")],
    labels=["Low", "Medium", "High"]
)

df[["median_income", "income_level"]].head(10)
````

````python
encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["income_level_encoded"] = encoder.fit_transform(
    df[["income_level"]]
).ravel()

df[["income_level", "income_level_encoded"]].head(10)
````

````python
from sklearn.preprocessing import LabelEncoder

weather_df = pd.read_csv("../00_datasets/seattle-weather.csv")

weather_df.head()
````

````python
le = LabelEncoder()

y_encoded = le.fit_transform(weather_df["weather"])

weather_df["weather_encoded"] = y_encoded

weather_df[["weather", "weather_encoded"]].head(10)
````

````python
print("Classes:", le.classes_)

for label, encoded_value in zip(
    le.classes_,
    le.transform(le.classes_)
):
    print(f"{label} -> {encoded_value}")
````

````python
print("Classes:", le.classes_)

for label, encoded_value in zip(
    le.classes_,
    le.transform(le.classes_)
):
    print(f"{label} -> {encoded_value}")
````

<h3 align="center" style="color:rgb(1, 126, 250);"><b>ORDINAL  ENCODING</b></h3>

> **Goal:** Convert categorical features into separate binary columns without implying an artificial ranking.

##### Example

| Weather | Rain | Snow | Sun |
| ------- | ---: | ---: | --: |
| Rain    |    1 |    0 |   0 |
| Snow    |    0 |    1 |   0 |
| Sun     |    0 |    0 |   1 |

##### Scikit-learn

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_encoded = encoder.fit_transform(X)
```

##### When to Use

- Categories have no natural order.
- Examples: City, weather type, product category.
- Useful for nominal categorical input features.

##### Remember

- Creates a separate binary column for each category by default.
- Avoids assigning arbitrary numerical rankings to categories.
- `handle_unknown="ignore"` handles unseen categories during transformation.
- Can increase the number of features when a column has many unique categories.



````python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

df = pd.read_csv("../00_datasets/seattle-weather.csv")

df.head()
````

````python
df["weather"].value_counts()
````

````python
X = df[["weather"]]

X.head()
````

````python
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_encoded = encoder.fit_transform(X)
````

````python
encoded_columns = encoder.get_feature_names_out(["weather"])

encoded_df = pd.DataFrame(
    X_encoded,
    columns=encoded_columns,
    index=X.index
)

encoded_df.head()
````

````python
print("Categories:", encoder.categories_)
print("Encoded shape:", encoded_df.shape)

encoded_df.sum()
````

````python
pd.concat([X, encoded_df], axis=1).head()
````
